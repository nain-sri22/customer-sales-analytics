```python
import random
from datetime import datetime, timedelta

random.seed(42)

products = ["Laptop", "Phone", "Headphones", "Tablet", "Keyboard"]
categories = ["Electronics", "Accessories"]
cities = ["Lucknow", "Delhi", "Mumbai", "Bangalore", "Pune", "Jaipur"]

data = []

start_date = datetime(2025, 1, 1)

for i in range(1, 101):
    product = random.choice(products)

    if product == "Laptop":
        price = random.choice([50000, 55000, 60000, 65000])
    elif product == "Phone":
        price = random.choice([20000, 25000, 30000, 35000])
    elif product == "Headphones":
        price = random.choice([1500, 2000, 2500, 3000])
    elif product == "Tablet":
        price = random.choice([15000, 18000, 20000, 25000])
    else:
        price = random.choice([800, 1000, 1200, 1500])

    quantity = random.randint(1, 5)
    sales = price * quantity

    data.append({
        "Order_ID": i,
        "Customer_ID": random.randint(1001, 1050),
        "Date": start_date + timedelta(days=random.randint(0, 364)),
        "Product": product,
        "Category": "Electronics" if product in ["Laptop", "Phone", "Tablet"] else "Accessories",
        "Quantity": quantity,
        "Price": price,
        "Sales": sales,
        "City": random.choice(cities)
    })

df = pd.DataFrame(data)

df.head()
```
